```python
#Determina todos los numeros impares hasta derterminado numero "a", despues los agrega uno a uno en una lista, cuando llega al ultimo los uno a uno.
import time

a= int(input("Dame un número ="))
cadena = range(0,a)
datos= []

for numero in cadena:                       #Aqui se establecen los numeros impares
   if numero%2 == 0:
      continue
   datos.append(numero)                     #Los numeros impares se agregan de uno en uno
   print(datos)
   time.sleep(0.035)

for i in range(1,len(datos)):               #Aqui se eliminan uno a uno
  datos.pop()
  print(datos)
  time.sleep(0.035)

```

```python
#Numeros primos - Determina los numeros primos en un rango a
import time

a= int(input("Dame un número ="))
inicio = time.time()
print(inicio)

for i in range(0,a):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1
        if conta>2: break

    if conta == 2:
        print(f'{i} es un primo')

fin = time.time()
print(fin)
print("t = ", (fin - inicio)*1000)
```
