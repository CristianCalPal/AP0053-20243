**Combinacion iterable**

```python
from re import X

#Dict Comprehensions

Nombre=["Alba", "Henry", "Sergio" ,"David","Haider","Daniel"]
Edad=[23,12,45,36,45]

#Es posible combianar dos listas para crear un diccionario utilizando la funcion zip

zipConocidos=zip(Nombre,Edad)
print(zipConocidos)                 #Devuelve la direccion de la iteracion

Conocidos=dict(zipConocidos)        #Convierte el zip en diccionario
print(Conocidos)

Celulares=("Samgsung","Xiaomi","HuaweI","Motorola")         #Se peuden combinar cualquier lista iterable
Procedencia=["Corea del sur","China", "China", "EEUU"]

dictCelular=dict(zip(Celulares,Procedencia))
print(dictCelular)

X=[1,2,3,4,5,6,7,8,9,0]
Y=[0,9,8,7,6,5,4,3,2,1]

zipCombinacion=zip(X,Y)

combinacion={key:value for key, value in zipCombinacion}
print(combinacion)
```
